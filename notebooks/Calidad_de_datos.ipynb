```python
# ====================================
# FUNCIÓN GENERAL DE CALIDAD DE DATOS
# ====================================

def analizar_dataframe(df, nombre="DataFrame"):
    
    print("=" * 60)
    print(f"ANÁLISIS DEL DATAFRAME: {nombre}")
    print("=" * 60)
    
    # Dimensiones
    print("\n📌 Dimensiones:")
    print(f"Filas: {df.shape[0]}")
    print(f"Columnas: {df.shape[1]}")
    
    # Tipos de datos
    print("\n📌 Tipos de datos:")
    print(df.dtypes.value_counts())
    
    # Valores nulos
    print("\n📌 Valores nulos:")
    
    nulos = df.isnull().sum()
    porcentaje_nulos = (nulos / len(df)) * 100
    
    tabla_nulos = pd.DataFrame({
        'Nulos': nulos,
        'Porcentaje': porcentaje_nulos
    })
    
    tabla_nulos = tabla_nulos[tabla_nulos['Nulos'] > 0]
    
    if len(tabla_nulos) == 0:
        print("No hay valores nulos.")
    else:
        print(tabla_nulos.sort_values(by='Porcentaje', ascending=False))
    
    # Duplicados
    print("\n📌 Duplicados:")
    print(df.duplicated().sum())
    
    # Memoria
    print("\n📌 Uso de memoria:")
    memoria = df.memory_usage(deep=True).sum() / 1024**2
    print(f"{memoria:.2f} MB")
    
    # Cardinalidad
    print("\n📌 Cardinalidad de variables:")
    
    cardinalidad = pd.DataFrame({
        'Variable': df.columns,
        'Valores únicos': [df[col].nunique() for col in df.columns]
    })
    
    print(cardinalidad.sort_values(by='Valores únicos', ascending=False))
    
    # Estadísticas numéricas
    print("\n📌 Estadísticas descriptivas:")
    display(df.describe())
    
    print("\n✅ Análisis completado.")
```
